```python
import numpy as np

class GridWorld:
    def __init__(self, size=5):
        self.size = size
        self.start = (0, 0)
        self.goal = (4, 4)
        self.state = self.start

    def reset(self):
        self.state = self.start
        return self.state

    def step(self, action):
        x, y = self.state
        if action == 0:   # up
            x = max(x - 1, 0)
        elif action == 1: # right
            y = min(y + 1, self.size - 1)
        elif action == 2: # down
            x = min(x + 1, self.size - 1)
        elif action == 3: # left
            y = max(y - 1, 0)

        self.state = (x, y)
        reward = 1 if self.state == self.goal else -0.01
        done = self.state == self.goal
        return self.state, reward, done
```

```python
import random

class QAgent:
    def __init__(self, env, alpha=0.1, gamma=0.9, epsilon=0.1):
        self.q = {}
        self.alpha = alpha
        self.gamma = gamma
        self.epsilon = epsilon
        self.env = env

    def get_q(self, state, action):
        return self.q.get((state, action), 0.0)

    def choose_action(self, state):
        if random.random() < self.epsilon:
            return random.choice([0,1,2,3])  # explore
        qs = [self.get_q(state, a) for a in [0,1,2,3]]
        return int(np.argmax(qs))  # exploit

    def update(self, state, action, reward, next_state, done):
        max_q_next = max([self.get_q(next_state, a) for a in [0,1,2,3]])
        td_target = reward + self.gamma * max_q_next * (1 - int(done))
        td_error = td_target - self.get_q(state, action)
        new_q = self.get_q(state, action) + self.alpha * td_error
        self.q[(state, action)] = new_q
```

```python
env = GridWorld()
agent = QAgent(env)

for episode in range(500):
    state = env.reset()
    done = False
    total_reward = 0
    while not done:
        action = agent.choose_action(state)
        next_state, reward, done = env.step(action)
        agent.update(state, action, reward, next_state, done)
        state = next_state
        total_reward += reward
    print(f"Episode {episode+1}: total reward = {total_reward:.2f}")
```

```python
actions = ['↑','→','↓','←']
for i in range(env.size):
    row = ''
    for j in range(env.size):
        state = (i, j)
        qs = [agent.get_q(state, a) for a in [0,1,2,3]]
        best = np.argmax(qs)
        row += actions[best] + ' '
    print(row)
```
